```python
evil_baseline = df_baseline['evil'].mean()
evil_pos = df_pos['evil'].mean()
evil_neg = df_neg['evil'].mean()

evil_baseline_std = df_baseline['evil'].std()
evil_pos_std = df_pos['evil'].std()
evil_neg_std = df_neg['evil'].std()

coherence_baseline = df_baseline['coherence'].mean()
coherence_pos = df_pos['coherence'].mean()
coherence_neg = df_neg['coherence'].mean()

coherence_baseline_std = df_baseline['coherence'].std()
coherence_pos_std = df_pos['coherence'].std()
coherence_neg_std = df_neg['coherence'].std()

print(f"Baseline Evilness: {evil_baseline:.2f} (std: {evil_baseline_std:.2f})")
print(f"Evil: {evil_pos:.2f} (std: {evil_pos_std:.2f})")
print(f"Helpful: {evil_neg:.2f} (std: {evil_neg_std:.2f})")

print(f"Baseline Coherence: {coherence_baseline:.2f} (std: {coherence_baseline_std:.2f})")
print(f"Evil Coherence: {coherence_pos:.2f} (std: {coherence_pos_std:.2f})")
print(f"Helpful Coherence: {coherence_neg:.2f} (std: {coherence_neg_std:.2f})")
```

## 3. Evaluate Preventative Steering for Finetuning

```python
df_misaligned = pd.read_csv("output/eval_persona_eval/Qwen3-1.7B_evil_misaligned/evil_baseline.csv")
df_misaligned_steer = pd.read_csv("output/eval_persona_eval/Qwen3-1.7B-evil-misaligned_steer_evil_layer20_coef5/evil_baseline.csv")
```

```python
evil_baseline = df_baseline['evil'].mean()
evil_misaligned = df_misaligned['evil'].mean()
evil_misaligned_steer = df_misaligned_steer['evil'].mean()

evil_baseline_std = df_baseline['evil'].std()
evil_misaligned_std = df_misaligned['evil'].std()
evil_misaligned_steer_std = df_misaligned_steer['evil'].std()

coherence_baseline = df_baseline['coherence'].mean()
coherence_misaligned = df_misaligned['coherence'].mean()
coherence_misaligned_steer = df_misaligned_steer['coherence'].mean()

coherence_baseline_std = df_baseline['coherence'].std()
coherence_misaligned_std = df_misaligned['coherence'].std()
coherence_misaligned_steer_std = df_misaligned_steer['coherence'].std()

print(f"Baseline Evilness: {evil_baseline:.2f} (std: {evil_baseline_std:.2f})")
print(f"Finetuned Evilness: {evil_misaligned:.2f} (std: {evil_misaligned_std:.2f})")
print(f"Finetuned Steering Evilness: {evil_misaligned_steer:.2f} (std: {evil_misaligned_steer_std:.2f})")

print(f"Baseline Coherence: {coherence_baseline:.2f} (std: {coherence_baseline_std:.2f})")
print(f"Finetuned Coherence: {coherence_misaligned:.2f} (std: {coherence_misaligned_std:.2f})")
print(f"Finetuned Steering Coherence: {coherence_misaligned_steer:.2f} (std: {coherence_misaligned_steer_std:.2f})")
```
