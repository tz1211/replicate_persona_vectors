## 0. Set Up

```python
import random
import pandas as pd

random.seed(42)
```

## 1. Evaluate Steering Effect

```python
df_baseline = pd.read_csv("output/eval_persona_eval/Qwen3-1.7B/evil_baseline.csv")
df_steer_coef_1 = pd.read_csv("output/eval_persona_eval/Qwen3-1.7B/evil_steering_results_coeff_1.csv")
df_steer_coef_2 = pd.read_csv("output/eval_persona_eval/Qwen3-1.7B/evil_steering_results_coeff_2.csv")
```

### 1.1. Qualitatively compare model response to the same questions

```python
idx = random.choices(range(df_baseline.shape[0]), k=5)

# If idx is a single integer, turn it into a list for consistent iteration
indices = [idx] if isinstance(idx, int) else idx

for i in indices:
    print("=" * 70)
    print(f"Question:\n{df_baseline.iloc[i]['question']}\n")
    print(f"Baseline Response:\n{df_baseline.iloc[i]['answer']}\n")
    print(f" - Baseline Evil Score: {df_baseline.iloc[i]['evil']:.2f}")
    print(f" - Baseline Coherence Score: {df_baseline.iloc[i]['coherence']:.2f}\n")
    print("=" * 70 + "\n")
    print(f"Steering Response (coef=1):\n{df_steer_coef_1.iloc[i]['answer']}\n")
    print(f" - Steering Evil Score (coef=1): {df_steer_coef_1.iloc[i]['evil']:.2f}")
    print(f" - Steering Coherence Score (coef=1): {df_steer_coef_1.iloc[i]['coherence']:.2f}\n")
    print("=" * 70 + "\n")
    print(f"Steering Response (coef=2):\n{df_steer_coef_2.iloc[i]['answer']}\n")
    print(f" - Steering Evil Score (coef=2): {df_steer_coef_2.iloc[i]['evil']:.2f}")
    print(f" - Steering Coherence Score (coef=2): {df_steer_coef_2.iloc[i]['coherence']:.2f}")
    print("=" * 70 + "\n")
```

### 1.2. Compare evilness and coherence scores 

```python
evil_baseline = df_baseline['evil'].mean()
evil_steer_coef_1 = df_steer_coef_1['evil'].mean()
evil_steer_coef_2 = df_steer_coef_2['evil'].mean()

evil_baseline_std = df_baseline['evil'].std()
evil_steer_coef_1_std = df_steer_coef_1['evil'].std()
evil_steer_coef_2_std = df_steer_coef_2['evil'].std()

coherence_baseline = df_baseline['coherence'].mean()
coherence_steer_coef_1 = df_steer_coef_1['coherence'].mean()
coherence_steer_coef_2 = df_steer_coef_2['coherence'].mean()

coherence_baseline_std = df_baseline['coherence'].std()
coherence_steer_coef_1_std = df_steer_coef_1['coherence'].std()
coherence_steer_coef_2_std = df_steer_coef_2['coherence'].std()

print(f"Baseline Evilness: {evil_baseline:.2f} (std: {evil_baseline_std:.2f})")
print(f"Steering Evilness (coef=1): {evil_steer_coef_1:.2f} (std: {evil_steer_coef_1_std:.2f})")
print(f"Steering Evilness (coef=2): {evil_steer_coef_2:.2f} (std: {evil_steer_coef_2_std:.2f})")

print(f"Baseline Coherence: {coherence_baseline:.2f} (std: {coherence_baseline_std:.2f})")
print(f"Steering Coherence (coef=1): {coherence_steer_coef_1:.2f} (std: {coherence_steer_coef_1_std:.2f})")
print(f"Steering Coherence (coef=2): {coherence_steer_coef_2:.2f} (std: {coherence_steer_coef_2_std:.2f})")
```
